```python
import json

losses = []
with open("/kaggle/working/runs/arm_D_500/metrics.jsonl") as f:
    for line in f:
        row = json.loads(line)
        if "step" in row:
            losses.append(row["loss"])

summary = {
    "n_steps": len(losses),
    "first_loss": losses[0],
    "last_loss": losses[-1],
    "min_loss": min(losses),
    "last_20_avg": sum(losses[-20:]) / len(losses[-20:]),
}
print(summary)

with open("/kaggle/working/metrics_train_arm_d_500.jsonl", "w") as f:
    f.write(json.dumps({"section": "train_arm_d_500", "results": summary}) + "\n")
```
