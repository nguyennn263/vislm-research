```python
import json

losses = []
with open("/kaggle/working/runs/arm_C_debug/metrics.jsonl") as f:
    for line in f:
        row = json.loads(line)
        if "step" in row:
            losses.append(row["loss"])

summary = {
    "n_steps": len(losses),
    "first_loss": losses[0],
    "last_loss": losses[-1],
    "min_loss": min(losses),
}
print(summary)

with open("/kaggle/working/metrics_train_arm_c_debug.jsonl", "w") as f:
    f.write(json.dumps({"section": "train_arm_c_debug", "results": summary}) + "\n")
```
